# Test Bot: Discord Webhook

Sends a simple greeting message to a Discord channel via an incoming webhook.

The webhook URL is read from the `DISCORD_FIJI_BOT` environment variable (backed by the repo secret of the same name) rather than hardcoded here.

```python
import os
import requests

WEBHOOK_URL = os.environ["DISCORD_FIJI_BOT"]
MESSAGE = os.environ.get("MESSAGE", "hi this is Fiji bot")

payload = {"content": MESSAGE}

response = requests.post(WEBHOOK_URL, json=payload)
print("Status code:", response.status_code)
print("Response body:", response.text)
response.raise_for_status()
```
